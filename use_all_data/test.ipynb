# Check sid duplicates

```python
import pandas as pd
import config
from utils import bagz_utils
```

```python
df = bagz_utils.read_parquet(config.META_W_ALL_SID) 
df.head(3)
```

```python
# Step 1: Extract the first three tokens
df['prefix_3'] = df['formatted_sid'].apply(lambda x: ' '.join(x.split()[:3]))

# Step 2: Count duplicates per prefix
group_counts = df.groupby('prefix_3').size().reset_index(name='count')

# Step 3: Sort groups by the number of duplicates (descending)
group_counts_sorted = group_counts.sort_values(by='count', ascending=False)

print(group_counts_sorted)
```

```python
# Suppose you want the group with prefix "A30 B233 C69"
prefix = "A35 B55 C50"

group_rows = df[df['prefix_3'] == prefix]

print(group_rows[['brand', 'title']])
```

```python
records = bagz_utils.read_record(config.USER_SEQUENCE)
records[0]
```

```python
from collections import Counter

# Step 1: Extract the last element of asin_sequence for each user
last_asins = [record['asin_sequence'][-1] for record in records]

# Step 2: Count frequency of each last ASIN
last_asin_counter = Counter(last_asins)

# Step 3: Get top-k most popular last ASINs
k = 10
top_k_last_asins = [asin for asin, _ in last_asin_counter.most_common(k)]
print("Top-k most popular last ASINs:", top_k_last_asins)

# Step 4: Compute recall@k
# Percentage of users whose last ASIN is in the top-k
hits = sum(1 for asin in last_asins if asin in top_k_last_asins)
recall_at_k = hits / len(last_asins)

print(f"Recall@{k}: {recall_at_k:.4f}")
```

# Upper bound

```python
import config
import bagz
import json
from collections import Counter
```

```python
data_reader = bagz.Reader(config.EVAL_DATA)
data = [json.loads(record.decode()) for record in data_reader]
print(len(data))

data[0]
```

```python
# Count how many times each target appears
target_counts = Counter(d['target'] for d in data)

# Get top-k most common targets
k = 5
# top_k_targets = target_counts.most_common(k)      # get item and count
top_k_targets = [t for t, _ in target_counts.most_common(k)]    # only get items


print(top_k_targets)
```

```python
# Step 3: Compute recall@k
hits = sum(1 for d in data if d['target'] in top_k_targets)
recall_at_k = hits / len(data)

print(f"Recall@{k}: {recall_at_k:.4f}")
```
